# Detect and Remove Outliers

```python
# import library 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv(
    r"D:\Intern Project on Taxi Trip Dataset\Taxi trip dataset\taxi_tripdata.csv",
    low_memory=False
)
```

```python
data.head(2)
```

```python
sns.boxplot(y='passenger_count', data=data)
```

```python
# Remove Outliers in Passenger_count Column 
# Plot Passenger_count in KDE  plot
plt.title("KDE  Plot of passenger_count")
sns.kdeplot(x='passenger_count', data=data)
```

```python
q1 = data['passenger_count'].quantile(0.25)
print(f"Q1 of passenger_count:{q1}")
```

```python
sns.boxplot(y='trip_distance', data=data)
```

```python
sns.boxplot(y='fare_amount', data=data)
```

```python
# Plot Sales in KDE  plot
plt.title("KDE  Plot of fare_amount ")
sns.kdeplot(x='fare_amount', data=data)
```

```python
q1 = data['fare_amount'].quantile(0.25)
print(f"Q1 of fare_amount:{q1}")
```

```python
q3 = data['fare_amount'].quantile(0.75)
print(f"Q3 of fare_amount:{q3}")
```

```python
# IQR , inter quantile range 
iqr = q3  - q1 
print(f"IQR of fare_amount is :{iqr}")
```

```python
# Lower Limit q1 -1.5 * iqr 
lower_limit= q1 - 1.5*iqr 
print(f"Lower Limit :{lower_limit}")
```

```python
# Upper Limit q3 + 1.5*iqr
upper_limit = q3 + 1.5 * iqr 
print(f"Upper Limit :{upper_limit}")
```

```python
sns.kdeplot(x='fare_amount', data=data)
plt.axvline(lower_limit, linestyle='--' , color='red', label ='Lower Limit' )
plt.axvline(upper_limit, linestyle='--', color='green', label='Upper Limit')
plt.show()
```

```python
# Remove Ouliters 
con1 = data['fare_amount']>=lower_limit
con2=data['fare_amount']<=upper_limit
```

```python
data[con1 & con2]
```

```python
df_iqr = data[con1 & con2]
```

```python
# Plot the Sales 
fig,axes=plt.subplots(1,2,figsize=(10,12))
sns.boxplot(y='fare_amount',data=data , ax=axes[0])
axes[0].set_title("Fare Amount Before Removing Outliers")
sns.boxplot(y='fare_amount',data=df_iqr , ax=axes[1])
axes[1].set_title("Fare Amount After Removing Outliers")
```

# Zscore Methods 

```python
from scipy import stats
```

```python
data['zscore'] = stats.zscore(data['fare_amount'])
```

```python
data
```

```python
data['zscore']  = abs(data['zscore'])
```

```python
data[data['zscore']<=3]
```

```python
df_zscore= data[data['zscore']<=3]
```

```python
# Plot the Sales 
fig,axes=plt.subplots(1,3,figsize=(14,12))
sns.boxplot(y='fare_amount',data=data , ax=axes[0])
axes[0].set_title("fare_amount Before Removing Outliers")
sns.boxplot(y='fare_amount',data=df_iqr , ax=axes[1])
axes[1].set_title("fare_amount After Removing Outliers (IQR)")
sns.boxplot(y='fare_amount',data=df_zscore , ax=axes[2])
axes[2].set_title("fare_amount After Removing Outliers (Zscore)")
plt.tight_layout()
```

```python
# STd and Mean 
sd =data['fare_amount'].std() 
mv = data['fare_amount'].mean()
```

```python
lower_bound = mv - 3 * sd
upper_bound = mv + 3 * sd
```

```python
sns.kdeplot(x='fare_amount', data=data)
plt.axvline(lower_bound)
plt.axvline(upper_bound)
```

```python
c1  = data['fare_amount']<=upper_bound
c2 = data['fare_amount']>=lower_bound
```

```python
data[c1 & c2]
```

```python
df_std_mean = data[c1 & c2]
```

```python
# Plot the Sales 
fig,axes=plt.subplots(1,4,figsize=(16,12))
sns.boxplot(y='fare_amount',data=data , ax=axes[0])
axes[0].set_title("fare_amount Before Removing Outliers")
sns.boxplot(y='fare_amount',data=df_iqr , ax=axes[1])
axes[1].set_title("fare_amount After Removing Outliers (IQR)")
sns.boxplot(y='fare_amount',data=df_zscore , ax=axes[2])
axes[2].set_title("fare_amount After Removing Outliers (Zscore)")
sns.boxplot(y='fare_amount',data=df_std_mean , ax=axes[3])
axes[3].set_title("fare_amount After Removing Outliers (Std Mean)")
plt.tight_layout()
```

```python
# Get Skew Value 
skew_all = data['fare_amount'].skew() 
skew_iqr = df_iqr['fare_amount'].skew()
skew_zscore= df_zscore['fare_amount'].skew()
skew_std_mean = df_std_mean['fare_amount'].skew()
```

```python
print(f"Skew of All Data:{skew_all}")
print(f"Skew of Data Removed with IQR: {skew_iqr}")
print(f"Skew of Data Removed with  Zscore :{skew_zscore}")
print(f"Skew of Data Remove with Std Mean: {skew_std_mean}")
```

```python
# Visual Inspection with KDE Plot 
# Plot the Sales 
fig,axes=plt.subplots(1,4,figsize=(16,5))
sns.kdeplot(x='fare_amount',data=data , ax=axes[0])
axes[0].set_title("fare_amount Before Removing Outliers")
sns.kdeplot(x='fare_amount',data=df_iqr , ax=axes[1])
axes[1].set_title("fare_amount After Removing Outliers (IQR)")
sns.kdeplot(x='fare_amount',data=df_zscore , ax=axes[2])
axes[2].set_title("fare_amount After Removing Outliers (Zscore)")
sns.kdeplot(x='fare_amount',data=df_std_mean , ax=axes[3])
axes[3].set_title("fare_amount After Removing Outliers (Std Mean)")
plt.tight_layout()
```

```python
# Scatter Plot 
# Plot the Sales 
fig,axes=plt.subplots(1,4,figsize=(16,5))
sns.scatterplot(x='fare_amount',y='total_amount', data=data , ax=axes[0])
axes[0].set_title("fare_amount Before Removing Outliers")
sns.scatterplot(x='fare_amount',y='total_amount', data=df_iqr , ax=axes[1])
axes[1].set_title("fare_amount After Removing Outliers (IQR)")
sns.scatterplot(x='fare_amount',y='total_amount',data=df_zscore , ax=axes[2])
axes[2].set_title("fare_amount After Removing Outliers (Zscore)")
sns.scatterplot(x='fare_amount',y='total_amount', data=df_std_mean , ax=axes[3])
axes[3].set_title("fare_amount After Removing Outliers (Std Mean)")
plt.tight_layout()
```
